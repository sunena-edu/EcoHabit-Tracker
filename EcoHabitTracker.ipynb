# EcoHabit Tracker & Sustainability Advisor

## Project Objective

This project evaluates a user's environmental habits and provides:

- Eco Score (0-100)
- Eco Level
- Personalized Recommendations
- Environmental Impact Summary
- Visual Chart Analysis

The goal is to encourage sustainable daily habits through technology.

```python
import matplotlib.pyplot as plt
```

```python
print("-" * 50)
print("🌱 ECOHABIT TRACKER & SUSTAINABILITY ADVISOR 🌱")
print("-" * 50)

print("\nAnswer the following questions honestly.\n")
```

```python
plastic_bottles = int(input("How many plastic bottles do you use per week? "))

public_transport = int(input("How many days per week do you use public transport (0-7)? "))

streaming_hours = int(input("How many hours of video streaming do you do per day? "))

reusable_bottle = input("Do you use a reusable water bottle? (yes/no): ").lower()

recycle = input("Do you recycle waste? (yes/no): ").lower()

lights = input("Do you switch off unused lights? (yes/no): ").lower()

short_showers = input("Do you keep showers short to save water? (yes/no): ").lower()

cloth_bag = input("Do you carry reusable shopping bags? (yes/no): ").lower()
```

```python
score = 50

# Plastic Bottles
if plastic_bottles <= 2:
    score += 15
elif plastic_bottles <= 5:
    score += 5
else:
    score -= 10

# Public Transport
score += public_transport * 2

# Streaming Hours
if streaming_hours <= 2:
    score += 10
elif streaming_hours > 6:
    score -= 10

# Reusable Bottle
if reusable_bottle == "yes":
    score += 15

# Recycling
if recycle == "yes":
    score += 15

# Lights
if lights == "yes":
    score += 10

# Water Conservation
if short_showers == "yes":
    score += 10

# Reusable Bags
if cloth_bag == "yes":
    score += 10
```

```python
score = max(0, min(score, 100))
```

```python
if score >= 80:
    level = "Eco Champion !!!! "

elif score >= 60:
    level = "Eco Explorer !!! "

elif score >= 40:
    level = "Eco Learner !! "

else:
    level = "Eco Beginner ! "
```

```python
print("\n" + "-" * 50)
print("RESULTS")
print("-" * 50)

print(f"\nEco Score: {score}/100")
print(f"Eco Level: {level}")
```

```python
print("\nPotential Positive Impact:")

if plastic_bottles > 5:
    print("- Could significantly reduce plastic waste annually.")

if public_transport < 3:
    print("- Could lower transportation-related emissions.")

if lights == "no":
    print("- Could reduce electricity consumption.")

if short_showers == "no":
    print("- Could conserve water resources.")

if cloth_bag == "no":
    print("- Could reduce plastic bag waste.")
```

```python
plastic_score = max(0, 20 - plastic_bottles)
transport_score = public_transport * 3
streaming_score = max(0, 20 - streaming_hours)
recycling_score = 20 if recycle == "yes" else 5
energy_score = 20 if lights == "yes" else 5
water_score = 20 if short_showers == "yes" else 5
```

```python
categories = ["Plastic","Transport","Streaming","Recycling","Energy","Water"]

scores = [plastic_score, transport_score, streaming_score, recycling_score, energy_score, water_score]

plt.figure(figsize=(8, 5))

plt.bar(categories, scores)

plt.title("Eco Habit Breakdown")

plt.xlabel("Categories")

plt.ylabel("Score")

plt.show()
```

```python
print("\n" + "-" * 50)
print("FINAL SUMMARY")
print("-" * 50)

print(f"\nEco Score: {score}/100")
print(f"Level: {level}")

if score >= 80:
    print("\nYou demonstrate highly sustainable habits. Keep it up.")

elif score >= 60:
    print("\nYou are environmentally conscious but still have room for improvement.")

elif score >= 40:
    print("\nYou have started developing sustainable habits.")

else:
    print("\nConsider adopting more environmentally friendly habits.")

print("\nThank you for using EcoHabit Tracker!")
```
